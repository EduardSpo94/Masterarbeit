```python
import laion_clap
import torch
import glob
import os
import re

# --- 1. ALL THREE PATHS ---
pfade = {
    "Synthetic Anomalies": "/content/drive/MyDrive/MasterProject/data/raw/mimii_dg/dev_bearing/bearing/train_anomaly_00",
    "Real Anomalies (Reference)": "/content/drive/MyDrive/MasterProject/data/raw/mimii_dg/dev_bearing/bearing/FAD_REF_REAL_ANOMALY_00",
    "Real Normal Data (Reference)": "/content/drive/MyDrive/MasterProject/data/raw/mimii_dg/dev_bearing/bearing/FAD_REF_REAL_NORMAL_00"
}

# --- 2. DECODER ---
def decodiere_dateiname_zu_caption(dateiname):
    if "anomaly" in dateiname:
        zustand = "with an anomaly"
    else:
        zustand = "normally"

    match = re.search(r'vel_(\d+)', dateiname)
    if match:
        geschwindigkeit = match.group(1)
    else:
        geschwindigkeit = "10" # Fallback

    return f"A bearing operating {zustand} at a rotation velocity of {geschwindigkeit} krpm. The sound contains factory noise at 12.0 dB SNR."

# --- 3. LOAD MODEL ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = laion_clap.CLAP_Module(enable_fusion=False, amodel='HTSAT-tiny')
model.load_ckpt()
model.to(device)

# --- 4. CALCULATION FOR ALL FOLDERS ---
ergebnisse = {}

with torch.no_grad():
    for name, pfad in pfade.items():
        audio_files = glob.glob(os.path.join(pfad, "*.wav"))
        scores = []

        if not audio_files:
            print(f"❌ Folder empty or not found: {name}")
            continue

        print(f"⏳ Calculating {name} ({len(audio_files)} files)...")

        for f in audio_files:
            basename = os.path.basename(f)
            prompt = decodiere_dateiname_zu_caption(basename)

            try:
                text_embed = model.get_text_embedding([prompt], use_tensor=True).to(device)
                text_norm = text_embed / text_embed.norm(dim=-1, keepdim=True)

                audio_embed = model.get_audio_embedding_from_filelist(x=[f], use_tensor=True).to(device)
                audio_norm = audio_embed / audio_embed.norm(dim=-1, keepdim=True)

                similarity = (audio_norm @ text_norm.T).item()
                scores.append(similarity)
            except:
                pass

        if scores:
            ergebnisse[name] = sum(scores) / len(scores)

# --- 5. FINAL TABLE ---
print("\n" + "=" * 55)
print("🏆 FINAL CLAP SCORES (SECTION 00)")
print("-" * 55)
for name, score in ergebnisse.items():
    print(f"{name:<30}: {score:.4f}")
print("=" * 55)
```

```python
import torch
import torchaudio
from hear21passt.base import get_basic_model
import glob
import os
import numpy as np
from torch.nn import functional as F
from scipy.stats import entropy

# --- 1. PATHS (SECTION 00) ---
PATH_SYNTH = "/content/drive/MyDrive/MasterProject/data/raw/mimii_dg/dev_bearing/bearing/train_anomaly_00"
PATH_REAL_ANOMALY = "/content/drive/MyDrive/MasterProject/data/raw/mimii_dg/dev_bearing/bearing/FAD_REF_REAL_ANOMALY_00"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# --- 2. LOAD MODEL ---
print("⏳ Loading PaSST classifier (AudioSet Expert)...")
model = get_basic_model(mode="logits")
model.eval()
model.to(device)

def extract_passt_probs(files, batch_size=32):
    """Runs audio files through PaSST and returns probabilities for 527 classes."""
    preds = []
    with torch.no_grad():
        for i in range(0, len(files), batch_size):
            batch_files = files[i:i+batch_size]
            batch_wavs = []

            for f in batch_files:
                wav, sr = torchaudio.load(f)
                # PaSST strictly requires 32kHz
                if sr != 32000:
                    resampler = torchaudio.transforms.Resample(sr, 32000)
                    wav = resampler(wav)

                # Exactly match 10 seconds (320,000 samples)
                if wav.shape[1] > 320000:
                    wav = wav[:, :320000]
                elif wav.shape[1] < 320000:
                    wav = F.pad(wav, (0, 320000 - wav.shape[1]))

                batch_wavs.append(wav)

            # Inference
            batch_tensor = torch.stack(batch_wavs).squeeze(1).to(device)
            logits = model(batch_tensor)
            probs = F.softmax(logits, dim=1) # Logits -> Probabilities
            preds.append(probs.cpu().numpy())

    return np.concatenate(preds, axis=0)

# --- 3. REFERENCE DISTRIBUTION (Real Anomalies) ---
print("\n🎧 Analyzing real anomalies (Reference)...")
files_real = glob.glob(os.path.join(PATH_REAL_ANOMALY, "*.wav"))
preds_real = extract_passt_probs(files_real)

# P_r: The empirical reality (Average model opinion on real anomalies)
P_r = np.mean(preds_real, axis=0)

# --- 4. SYNTHETIC DISTRIBUTION (Generated Anomalies) ---
print("🎧 Analyzing synthetic anomalies...")
files_synth = glob.glob(os.path.join(PATH_SYNTH, "*.wav"))
preds_synth = extract_passt_probs(files_synth)

# P_g: The approximated distribution (Average opinion on synthetic data)
P_g = np.mean(preds_synth, axis=0)

# --- 5. CALCULATION OF METRICS (According to Thesis) ---

# A) Inception Score (Equation 3) - Computed only on generated data
# Computes the deviation of each individual sample from the overall distribution
kl_divs_is = np.sum(preds_synth * (np.log(preds_synth + 1e-16) - np.log(P_g + 1e-16)), axis=1)
inception_score = np.exp(np.mean(kl_divs_is))

# B) KL-Divergence (Equation 2) - D_KL(P_g || P_r)
# Compares the information deviation between synthetic and reference data
kl_divergence = entropy(pk=P_g, qk=P_r)

# --- 6. OUTPUT ---
print("\n" + "=" * 45)
print("🏆 FINAL SEMANTIC METRICS (SECTION 00)")
print("-" * 45)
print(f"Inception Score (IS): {inception_score:.4f} (Higher is better)")
print(f"KL-Divergence (DKL):  {kl_divergence:.4f} (Lower is better)")
print("=" * 45)
```
